## Exploring and preprocessing the Families in the Wild dataset

Dataset is taken from [here](https://github.com/Zhangxm520/RFIW2021/blob/master/Track1/dataset.zip), the photos are already cropped and realigned (couldn't find the original dataset)

```python
import os
from os.path import join

base_path = r"C:\Users\tre_i\Downloads\dataset"

os.listdir(base_path)
```

```python
files = []
root = join(base_path, join('Train', 'train-faces'))
for dir in os.listdir(root):
  for folder in os.listdir(os.path.join(root, dir)):
    if os.path.isdir(os.path.join(root, dir, folder)):
      for file in os.listdir(os.path.join(root, dir, folder)):
        files.append(os.path.join(root, dir, folder, file))
```

```python
root = join(base_path, join('Validation', 'val-faces'))
for dir in os.listdir(root):
  for folder in os.listdir(os.path.join(root, dir)):
    for file in os.listdir(os.path.join(root, dir, folder)):
      files.append(os.path.join(root, dir, folder, file))
```

```python
root = join(base_path, join('Test', 'test-faces'))
for file in os.listdir(root):
  files.append(os.path.join(root, file))
```

```python
%%time
from src.embedding_generation import get_files_faces
from pickle import dump


buffalo_sc_embeddings = get_files_faces(files, padding_params=True)

faces_info = {}
for k, v in buffalo_sc_embeddings.items():
  if v is not None:
    faces_info[k] = {
        "embedding": v["embedding"],
        "det_score": v["det_score"]
    }
  else:
    faces_info[k] = None

with open('files/FIW_buffalo_sc_faces_info.pkl', 'wb') as f:
  dump(faces_info, f)
```
