```python
import json


with open('files/FIW_family_members.json') as f:
    families_json = json.load(f)
```

```python
import pandas as pd

families_subset = pd.read_csv('files/family_subset.csv', index_col=0)
```

```python
test_families_set = set(families_subset[families_subset['subset'] == 'test'].index)
```

```python
import random

test_families = {}

json_new = []
for person in families_json:
    family_id = person["family_id"]
    if family_id in test_families_set:
        test_families[family_id] = test_families.get(family_id, []) + [person]
```

```python
from copy import deepcopy

for family_id, people in test_families.items():
    people_to_choose = [el for el in people if 'additional' not in el["person_id"]]
    person_to_assign = random.choice(people_to_choose)
    other_people = deepcopy(people)
    other_people.remove(person_to_assign)
    person_to_assign['family_id'] = family_id + '_split'
    person_to_assign['relatives'] = []
    json_new.append(person_to_assign)
    for person in other_people:
        relatives = []
        for rel in person["relatives"]:
            if rel['person_id'] != person_to_assign["person_id"]:
                relatives.append(rel)
        person["relatives"] = relatives
        json_new.append(person)
```

```python
with open('files/FIW_test_for_simulation_family_members.json', 'w') as f:
    json.dump(json_new, f, indent=4)
```
