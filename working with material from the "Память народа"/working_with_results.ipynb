```python
import json

with open("res_dict_1.json", encoding="utf-8") as f:
    res = json.load(f)
```

```python
for el in res.values():
    info = el["information"]
    info = info[info.find("Дата рождения") + len("Дата рождения:</b>"):]
    info = info[:info.find("div") - 2]
    while len(info) > 0 and info[-1] in {" ", "\n", "\t"}:
        info = info[:-1]
    while len(info) > 0 and info[0] in {" ", "\n", "\t"}:
        info = info[1:]
    el["birth_date"] = info
    print(info)
```

```python
for el in res.values():
    info = el["information"]
    info = info[info.find("Место рождения") + len("Место рождения:</b>"):]
    info = info[:info.find("div") - 2]
    while len(info) > 0 and info[-1] in {" ", "\n", "\t"}:
        info = info[:-1]
    while len(info) > 0 and info[0] in {" ", "\n", "\t"}:
        info = info[1:]
    el["birth_place"] = info
    del el["information"]
    print(info)
```
