```python
import insightface
from insightface.app.common import Face
from insightface.model_zoo import model_zoo


det_model_path = 'det_10g.onnx'
rec_model_path = 'w600k_r50.onnx'
```

```python
det_model = model_zoo.get_model(f'{det_model_path}')
rec_model = model_zoo.get_model(f'{rec_model_path}')

det_model.prepare(ctx_id=0, input_size=(640, 640))
```

Let's set the detection threshold low at the minimum.

```python
det_model.det_thresh = 0
det_model.det_thresh
```

Padding should be unnecessary in the case of this dataset

```python
import cv2


##
def get_embedding_from_img_path(img_path):
    rgb_img = cv2.cvtColor(cv2.imread(img_path), cv2.COLOR_BGR2RGB)
    bboxes, kpss = det_model.detect(rgb_img)
    if bboxes.shape[0] == 0:
        return None
    i = 0
    bbox = bboxes[i, 0:4]
    det_score = bboxes[i, 4]

    kps = None
    if kpss is not None:
        kps = kpss[i]

    face = Face(bbox=bbox, kps=kps, det_score=det_score)
    embedding = rec_model.get(rgb_img, face)
    return embedding
```

```python
%%time

embedding_1 = get_embedding_from_img_path('/content/SiblingsDB/DBs/HQf/1/_DSC5755_1.jpg')
```

```python
paths = []
for dataset in ['HQf', 'HQfp', 'HQfps', 'LQf']:
  path = f'/content/SiblingsDB/DBs/{dataset}'
  if dataset == 'LQf':
    for file in os.listdir(path):
      if ".jpg" in file:
        paths.append(os.path.join(path, file))
  else:
    for folder in os.listdir(path):
      for file in os.listdir(os.path.join(path, folder)):
        if ".jpg" in file:
          paths.append(os.path.join(path, folder, file))
paths
```

```python
from tqdm import tqdm


face_dict = {}
for path in tqdm(paths):
  image = cv2.imread(path)
  faces = app.get(image)
  file_name = '/'.join(path.split('/')[2:])
  if len(faces) > 0:
    face_dict[file_name] = faces[0]
  else:
    face_dict[file_name] = None
```

```python
[k for  k, v in face_dict.items() if v is None]
```

```python
list(face_dict.values())[0].keys()
```

```python
faces_info = {}
for k, v in face_dict.items():
  if v is not None:
    faces_info[k] = {
        "embedding": v["embedding"],
        "det_score": v["det_score"]
    }
  else:
    faces_info[k] = None
```

```python
from pickle import dump


with open('SiblingsDB_faces_info.pkl', 'wb') as f:
  dump(faces_info, f)
```
