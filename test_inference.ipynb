Metrics on all experiments

```python
test_df["not_relative_probability"] = test_proba[:, not_relative_index]
test_df["child_probability"] = test_proba[:, child_index]
test_df["parent_probability"] = test_proba[:, parent_index]
test_df["sibling_probability"] = test_proba[:, sibling_index]
```

```python
from pickle import load


with open('files/FIW_test_families.pkl', 'rb') as pickle_file:
    families = load(pickle_file)
```

```python
top_1 = []
top_3 = []
top_5 = []

top_1_family = []
top_3_family = []
top_5_family = []

dfs = []

for assignment_task, df in test_df.groupby(by='assignment_task'):
    df['df_index'] = df.index
    df = df[['child_probability', 'parent_probability', 'sibling_probability', 'kinship', 'not_relative_probability', 'df_index']]
    df = df.melt(id_vars=['kinship', 'not_relative_probability', 'df_index'],
    value_vars=['child_probability', 'parent_probability', 'sibling_probability'],
    var_name="score_type",
    value_name="probability")


    for index in df.index:
        row = df.loc[index]
        kinship = row['kinship']
        score_type = row['score_type']
        idx = row['df_index']
        family_id = idx[3]
        correct_relative = (kinship == 'sibling' and score_type == 'sibling_probability') or (kinship == 'parent' and score_type == 'parent_probability') or (kinship == 'child' and score_type == 'child_probability')
        correct_family = idx[1] == family_id
        correct_guess = correct_relative and correct_family

        df.loc[index, 'correct_guess'] = correct_guess
        df.loc[index, 'correct_family'] = correct_family

        if kinship == 'parent':
            person = families[family_id].members[idx[2]]
            person_to_assign = families[idx[1]].members[idx[0]]
            if person_to_assign.gender == "Male" and len(person.father) > 0 and not person_to_assign in person.father:
                df.loc[index, "possible"] = False
            elif person_to_assign.gender == "Female" and len(person.mother) > 0 and not person_to_assign in person.mother:
                df.loc[index, "possible"] = False
            else:
                df.loc[index, "possible"] = True
        else:
            df.loc[index, "possible"] = True

    df = df[df['possible']]

    df = df.sort_values(by=['probability'], ascending=False)

    top_1.append(df.iloc[0]["correct_guess"])
    top_3.append(df.iloc[:3]["correct_guess"].any())
    top_5.append(df.iloc[:5]["correct_guess"].any())

    top_1_family.append(df.iloc[0]["correct_family"])
    top_3_family.append(df.iloc[:3]["correct_family"].any())
    top_5_family.append(df.iloc[:5]["correct_family"].any())

    dfs.append(df)
```

```python
print(f"Top-1: {sum(top_1) / len(top_1)}")
print(f"Top-3: {sum(top_3) / len(top_3)}")
print(f"Top-5: {sum(top_5) / len(top_5)}")

print(f"Top-1 family: {sum(top_1_family) / len(top_1_family)}")
print(f"Top-3 family: {sum(top_3_family) / len(top_3_family)}")
print(f"Top-5 family: {sum(top_5_family) / len(top_5_family)}")
```

```python
from tqdm import tqdm
import ast
from src.short_list_selection import select_closest


all_people = []
for k, v in families.items():
    all_people += list(v.members.values())

in_top_guesses = []
for el in tqdm(set(test_df['assignment_task'])):
    person_id, family_id = ast.literal_eval(el)
    family = families[family_id]
    person = family.members[person_id]
    family_excluding = family.create_family_excluding_a_person(person_id)#
    all_other_people = [p for p in all_people if p.family_id != family_id] + [person]
    top_choices = select_closest(family_excluding, all_other_people, 10)
    in_top_guesses.append(person_id in [el[2] for el in top_choices])
```
