```python
folder = "/mnt/c/Users/tre_i/Downloads/images"
files = os.listdir(folder)
```

```python
import cv2
```

```python
from tqdm import tqdm


face_dict = {}
for el in tqdm(files):
  
  path = os.path.join(folder, el)
  file_name = el[:el.rfind(".")]
  try:
      image = cv2.imread(path)
      faces = app.get(image)
      
      if len(faces) > 0:
        face_dict[file_name] = faces[0]
      else:
        face_dict[file_name] = None
  except:
      face_dict[file_name] = None
```

```python
len(face_dict)
```

```python
len([el for el in face_dict if face_dict[el] is None])
```

```python
faces_info = {}
for k, v in face_dict.items():
  if v is not None:
    faces_info[k] = {
        "embedding": v["embedding"],
        "det_score": v["det_score"]
    }
  else:
    faces_info[k] = None
```

```python
from pickle import dump


with open('Wikipedia_buffalo_sc_faces_info.pkl', 'wb') as f:
  dump(faces_info, f)
```
