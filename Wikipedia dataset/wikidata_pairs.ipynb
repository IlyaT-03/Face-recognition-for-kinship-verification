```python
import pandas as pd
```

```python
df = pd.read_csv('query.csv')
df
```

```python
df["person"] = df["person"].apply(lambda x: x.split('/')[-1])
df["child"] = df["child"].apply(lambda x: x.split('/')[-1])
df
```

```python
df["personGender"] = df["personGender"].apply(lambda x: x.split('/')[-1])
df["childGender"] = df["childGender"].apply(lambda x: x.split('/')[-1])
gender_dict = {"Q6581097": "male", "Q6581072": "female"}
df["personGender"] = df["personGender"].apply(lambda x: gender_dict[x] if x in gender_dict else x)
df["childGender"] = df["childGender"].apply(lambda x: gender_dict[x] if x in gender_dict else x)
df
```

```python
len(set(df["person"])), len(set(df["child"])), len(set(df["person"]).union(set(df["child"]))), len(set(df["person"]).intersection(set(df["child"])))
```

```python
people_list = list(set(df["person"]).union(set(df["child"])))
```

```python
person_num_dict = {}
for i in range(len(people_list)):
  person_num_dict[people_list[i]] = i
```

```python
import numpy as np


graph = np.zeros((len(people_list), len(people_list)))
for i in range(len(df)):
  graph[person_num_dict[df["child"][i]]][person_num_dict[df["person"][i]]] = 1
```

```python
from scipy.sparse import csr_array
from scipy.sparse.csgraph import connected_components

graph = csr_array(graph)
n_components, labels = connected_components(csgraph=graph, directed=True, return_labels=True)
```

```python
n_components
```

```python
labels
```

```python
id_to_family_num = {person_id: family_num for person_id, family_num in zip(people_list, labels)}
```

```python
families = {}
for i in range(n_components):
  family_index_list = np.where(labels == i)[0]
  family_ids_list = [people_list[index] for index in family_index_list]
  family_graph = graph[family_index_list][:, family_index_list].toarray()
  family_graph_df = pd.DataFrame(family_graph, index=family_ids_list, columns=family_ids_list).astype(int)
  families[i] = {"ids": family_ids_list, "graph": family_graph_df}
```

```python
child_parents_dict = {}
for parent, child in zip(df["person"], df["child"]):
  if child not in child_parents_dict:
    child_parents_dict[child] = {parent}
  else:
    child_parents_dict[child].add(parent)
```

```python
person1_list = []
person2_list = []
relationship_list = []
family_list = []
for child in child_parents_dict:
  for parent in child_parents_dict[child]:
    person1_list.append(parent)
    person2_list.append(child)
    relationship_list.append("parent - child")
    family_list.append(id_to_family_num[parent])
    if parent in child_parents_dict:
      for grandparent in child_parents_dict[parent]:
        person1_list.append(grandparent)
        person2_list.append(child)
        relationship_list.append("grandparent - child")
        family_list.append(id_to_family_num[grandparent])
```

```python
from tqdm import tqdm

for child_1 in tqdm(child_parents_dict):
  family_id = id_to_family_num[child_1]
  for child_2 in families[family_id]["ids"]:
    if child_2 in child_parents_dict and child_1 != child_2:
      if child_parents_dict[child_1] == child_parents_dict[child_2]:
        person1_list.append(child_1)
        person2_list.append(child_2)
        relationship_list.append("siblings")
        family_list.append(id_to_family_num[child_1])
```

```python
pairs_df = pd.DataFrame({"person1": person1_list, "person2": person2_list, "relationship": relationship_list, "family": family_list})
pairs_df
```

```python
person_to_pictures = {}
for i in range(len(df)):
  if df["person"][i] not in person_to_pictures:
    person_to_pictures[df["person"][i]] = [df["personImage"][i]]
  else:
    person_to_pictures[df["person"][i]].append(df["personImage"][i])
  if df["child"][i] not in person_to_pictures:
    person_to_pictures[df["child"][i]] = [df["childImage"][i]]
  else:
    person_to_pictures[df["child"][i]].append(df["childImage"][i])
```

```python
picture_pairs_df = pd.DataFrame(columns=["person1", "person2", "image1", "image2", "relationship", "family"])
for i in tqdm(range(len(pairs_df))):
  for picture1 in person_to_pictures[pairs_df["person1"][i]]:
    for picture2 in person_to_pictures[pairs_df["person2"][i]]:
      picture_pairs_df.loc[len(picture_pairs_df)] = [pairs_df["person1"][i], pairs_df["person2"][i], picture1, picture2, pairs_df["relationship"][i], pairs_df["family"][i]]
```

```python
train_picture_pairs_df = picture_pairs_df[picture_pairs_df["family"].isin(train_families)]
val_picture_pairs_df = picture_pairs_df[picture_pairs_df["family"].isin(val_families)]
test_picture_pairs_df = picture_pairs_df[picture_pairs_df["family"].isin(test_families)]
```

```python
picture_to_person = {}
for i in range(len(df)):
  picture_to_person[df["personImage"][i]] = df["person"][i]
  picture_to_person[df["childImage"][i]] = df["child"][i]
```

```python
train_pictures = [el for el in picture_to_person if id_to_family_num[picture_to_person[el]] in set(train_families)]
val_pictures = [el for el in picture_to_person if id_to_family_num[picture_to_person[el]] in set(val_families)]
test_pictures = [el for el in picture_to_person if id_to_family_num[picture_to_person[el]] in set(test_families)]
```

```python
counter = 0

train_picture_pairs_non_relatives_df = pd.DataFrame(columns=["person1", "person2", "image1", "image2", "relationship"])
while len(train_picture_pairs_non_relatives_df) < len(train_picture_pairs_df):
  i = np.random.randint(0, len(train_pictures))
  j = np.random.randint(0, len(train_pictures))
  if i != j:
    family_1 = id_to_family_num[picture_to_person[train_pictures[i]]]
    family_2 = id_to_family_num[picture_to_person[train_pictures[j]]]
    if family_1 != family_2:
      row = [picture_to_person[train_pictures[i]], picture_to_person[train_pictures[j]], train_pictures[i], train_pictures[j], None]
      train_picture_pairs_non_relatives_df.loc[len(train_picture_pairs_non_relatives_df)] = row
  counter += 1
  if counter % 1000 == 0:
    print(counter)
```
